```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
n_sample = 1500
no = 825
yes = 495
no_answer = 180
percentage=['55','33','12']
```

```python
fig, ax = plt.subplots()
bars = ax.bar(['NO','YES','NO ANSWER'], [825,495,180])
bars[0].set_color('g')
bars[1].set_color('r')
bars[2].set_color('tab:grey')
ax.set_ylabel('Votes')
ticks_loc = np.arange(0, n_sample+100, step=100)
ax.set_yticks(ticks_loc)
ax.set_ylim(0,n_sample+100)
ax.grid(False)

fig.set_size_inches(10,5)
fig.set_dpi(100)

ax.set_title('Do you want to send weapons to Ukraine?')
i = 0
for b in bars:
    height = b.get_height()
    ax.text(x=b.get_x() + b.get_width() / 2, y=height+.10,s=percentage[i]+'%',ha='center')
    i+=1

    
plt.savefig('survey.png')    
plt.show()
```
