```python
import chess
import chess.pgn
import pandas as pd
import numpy as np
```

```python
pgn = open("moves.pgn")
```

Initializing lists to store the name of white, the name of black, the date and result

```python
white = []
black = []
date = []
result = []
```

Iterate through each game in the PGN file

```python
while True:
    game= chess.pgn.read_game(pgn)
    if game is not None:
        result.append(game.headers['Result'])
        white.append(game.headers['White'])
        black.append(game.headers['Black'])
        date.append(game.headers['Date'])    
    else: 
        break    
```

```python
print(len(white))
print(len(black))
print(len(date))
print(len(result))
```

```python
df = pd.DataFrame(list(zip(white,black,result,date)),
                 columns=['WHITE','BLACK','RESULT','DATE'])
```

Collecting the moves of each game in algebric notation

```python
alg_not = []
```

```python
while True:
    game= chess.pgn.read_game(pgn)
    if game is not None:
        board = game.board()
        algebric_moves=[]
        for move in game.mainline_moves():
            alg = board.san(move)
            algebric_moves.append(alg)
            board.push(move)
        if not algebric_moves:                #the algebric_moves list is used to check if the are no moves in that game
            alg_not.append(np.nan)
        else:
            alg_not.append(algebric_moves)   
    else: 
        break
```

```python
len(alg_not)
```

Since the idea is to create a dataframe, the lists are sliced in order to have the same length

```python
a = white[:len(alg_not)]
```

```python
b = black[:len(alg_not)]
```

```python
c = result[:len(alg_not)]
```

```python
d = date[:len(alg_not)]
```

```python
len(d)
```

```python
df2 = pd.DataFrame(list(zip(a,b,c,d,alg_not)),
                 columns=['WHITE','BLACK','RESULT','DATE','MOVES'])
```

```python
df2
```

Collecting the elo of the white and black player

```python
w_elo = []
b_elo = []
```

```python
while True:
    game= chess.pgn.read_game(pgn)
    if game is not None:
        try:
            b_elo.append(game.headers['BlackElo'])
        except:
            b_elo.append(np.nan)
            continue
        try:
            w_elo.append(game.headers['WhiteElo'])
        except:
            w_elo.append(np.nan)
            continue;
    else: 
        break
```

```python
len(w_elo)
```

```python
len(df2)
```

```python
sliced_w_elo = w_elo[:len(df2)]
sliced_b_elo = b_elo[:len(df2)]
```

```python
len(sliced_w_elo)
```

```python
len(sliced_b_elo)
```

```python
df['B_ELO'] = sliced_b_elo
```
