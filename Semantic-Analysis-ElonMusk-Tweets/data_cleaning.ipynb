```python
import pandas as pd
import json
```

Helping Functions

```python
def split_after(my_list,separator):    
    for i,s in enumerate(my_list):
        if separator in s:
            my_list[i] = s.split(separator,1)[1] 
        
def split_before(my_list,separator):    
    for i,s in enumerate(my_list):
        if separator in s:
            my_list[i] = s.split(separator,1)[0]
        
def remove_character(my_list,to_rem):
    for i,s in enumerate(my_list):
        if to_rem in s:
            my_list[i] = s.replace(to_rem,'')
```

Fill missing values in the mention coloumn with 'No mention'

```python
df = pd.read_csv('elon.csv')
df=df.drop('Unnamed: 0', axis=1)
```

```python
#clean mention coloumn
df['Mention']=df['Mention'].fillna('No mention')
```

```python
df
```

Now we want to clean the Mention column in order to have just the username of the mentioned user

```python
mention_arr = df['Mention'].to_numpy()
```

```python
split_before(mention_arr,',')
```

```python
split_after(mention_arr,'=')
```

```python
mention_arr[0]
```

```python
remove_character(mention_arr,'\'')
```

Adding the @ symbol to keep the shape of the followers and following

```python
for i,s in enumerate(mention_arr):
    if s != 'No mention':
        mention_arr[i] ="@"+s
```

```python
df=df.drop('Mention', axis=1)
```

```python
df['Mention'] = mention_arr.tolist()
```

```python
df
```

```python
df=df.drop('Date', axis=1)
```

```python
df.to_csv('elon-cleaned.csv')
```
